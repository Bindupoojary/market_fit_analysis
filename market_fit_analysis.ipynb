```python
import pandas as pd

aqi = pd.read_csv("day-wise-state-wise-air-quality-index-aqi-of-major-cities-and-towns-in-india.csv")

disease = pd.read_csv(
    "master-data-state-district-and-disease-wise-cases-and-death-reported-due-to-outbreak-of-diseases-as-per-weekly-reports-under-idsp.csv",
    encoding="latin1"
)

vehicles = pd.read_csv("master-data-state-vehicle-class-and-fuel-type-wise-total-number-of-vehicles-registered-in-each-month-in-india.csv")

population = pd.read_excel("population-projection-of-india-state-and-gender-wise-yearly-projected-urban-population-2011-2036.xlsx")


# -------------------------------
# 2. A function to clean column names
# -------------------------------
def clean_columns(df, df_name):
    print(f"\n--- Cleaning columns for: {df_name} ---")

    old_cols = df.columns.tolist()

    df.columns = (
        df.columns
        .str.strip()
        .str.lower()
        .str.replace(" ", "_")
        .str.replace("/", "_")
        .str.replace("-", "_")
    )

    new_cols = df.columns.tolist()

    mapping = {old: new for old, new in zip(old_cols, new_cols) if old != new}

    if mapping:
        print("Column name changes:")
        for old, new in mapping.items():
            print(f"  '{old}' → '{new}'")
    else:
        print("No column names changed.")

    return df


# -------------------------------
# 3. Clean AQI dataset
# -------------------------------
print("\n=== Cleaning AQI Dataset ===")

aqi["number_of_monitoring_stations"] = aqi["number_of_monitoring_stations"].fillna(0)
aqi["prominent_pollutants"] = aqi["prominent_pollutants"].fillna("Unknown")

before = len(aqi)
aqi = aqi.drop_duplicates()
print(f"Removed {before - len(aqi)} duplicates from AQI")

aqi = clean_columns(aqi, "AQI")


# -------------------------------
# 4. Clean Disease dataset
# -------------------------------
print("\n=== Cleaning Disease Dataset ===")

disease["outbreak_starting_date"] = pd.to_datetime(disease["outbreak_starting_date"], errors="coerce", dayfirst=True)
disease["reporting_date"] = pd.to_datetime(disease["reporting_date"], errors="coerce", dayfirst=True)

disease["state"] = disease["state"].ffill()
disease["district"] = disease["district"].ffill()
disease["disease_illness_name"] = disease["disease_illness_name"].ffill()

disease["cases"] = disease["cases"].fillna(0)
disease["deaths"] = disease["deaths"].fillna(0)

before = len(disease)
disease = disease.drop_duplicates()
print(f"Removed {before - len(disease)} duplicates from Disease dataset")

disease = clean_columns(disease, "Disease")


# -------------------------------
# 5. Clean Vehicles dataset
# -------------------------------
print("\n=== Cleaning Vehicles Dataset ===")

before = len(vehicles)
vehicles = vehicles.drop_duplicates()
print(f"Removed {before - len(vehicles)} duplicates from Vehicles dataset")

vehicles = clean_columns(vehicles, "Vehicles")


# -------------------------------
# 6. Clean Population dataset
# -------------------------------
print("\n=== Cleaning Population Dataset ===")

before = len(population)
population = population.drop_duplicates()
print(f"Removed {before - len(population)} duplicates from Population dataset")

population = clean_columns(population, "Population")


# -------------------------------
# 7. Save all cleaned versions
# -------------------------------

aqi.to_csv("cleaned_aqi.csv", index=False)
disease.to_csv("cleaned_disease.csv", index=False)
vehicles.to_csv("cleaned_vehicles.csv", index=False)
population.to_excel("cleaned_population.xlsx", index=False)

print("\n\n All datasets cleaned and saved successfully!")
print("Files saved as: cleaned_aqi.csv, cleaned_disease.csv, cleaned_vehicles.csv, cleaned_population.xlsx")

```
